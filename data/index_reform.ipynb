```python
import pandas as pd
```

## OIQA

```python
def oiqa_process(idx):
    if len(idx.split('.'))==1 and len(idx.split('/'))==1:
        return idx
    return f"img{int(idx.split('/')[-1].split('.')[0])}"
def cviq_process(idx):
    if type(idx)==int or (len(idx.split('.'))==1 and len(idx.split('/'))==1):
        return idx
    return f"{int(idx.split('/')[-1].split('.')[0]):03d}"
def odi_process(idx):
    if len(idx.split('.'))==1 and len(idx.split('/'))==1:
        return idx
    return idx.split('/')[-1].split('.')[0]
def mvaqd_process(idx):
    if type(idx)==int or (len(idx.split('.'))==1 and len(idx.split('/'))==1):
        return idx
    return f"{int(idx.split('/')[-1].split('.')[0])}"
def jufe_process(idx):
    return idx.split('.')[0]
def process_file(func, path):
    df_train = pd.read_csv(path,header=None,names = ['img','mos'])
    df_train['img'] = df_train['img'].apply(func)
    df_train.to_csv(path,header=None, index=False)
def process_oiq(func, path):
    df_train = pd.read_csv(path,header=None,names = ['img','mos','numfd'])
    df_train['img'] = df_train['img'].apply(func)
    df_without_numfd = df_train.drop(columns=['numfd'])
    df_without_numfd.to_csv(path,header=None, index=False)
```

```python
for i in range(1,6):
    for t in ['test','train']:
        process_file(oiqa_process,f'OIQA/OIQA_{t}c_{i}.csv')
    process_file(oiqa_process, f'OIQA/OIQA.csv')
    for dis in ['GB','JPEG','JP2K','WN']:
        process_file(oiqa_process,f'OIQA/dis_type/OIQA_testc_{dis}{i}.csv')
        
    
```

```python
for i in range(1,6):
    for t in ['test','train']:
        process_file(cviq_process,f'CVIQ/CVIQ_{t}c_{i}.csv')
    process_file(cviq_process,f'CVIQ/CVIQ.csv')
    for dis in ['AVC','JPEG','HEVC']:
        process_file(cviq_process,f'CVIQ/dis_type/CVIQ_testc_{dis}{i}.csv')
```

```python
for i in range(1,6):
    for t in ['test','train']:
        process_file(odi_process,f'IQA-ODI/IQA-ODI_{t}c_{i}.csv')
    process_file(odi_process,f'IQA-ODI/IQA-ODI.csv')
    process_file(odi_process,f'IQA-ODI/IQA-ODI_1080.csv')
    for dis in ['CMP','CPP','ERP','ISP','OHP']:
        process_file(odi_process,f'IQA-ODI/dis_type/IQA-ODI_testc_{dis}{i}.csv')
```

```python
for i in range(1,6):
    for t in ['test','train']:
        process_file(mvaqd_process,f'MVAQD/MVAQD_{t}c_{i}.csv')
    for dis in ['GB','HEVC','JP2K','WN','JPEG']:
        process_file(mvaqd_process,f'MVAQD/dis_type/MVAQD_testc_{dis}{i}.csv')
        
```

```python
for i in range(1,6):
    for t in ['test','train']:
        process_file(jufe_process,f'JUFE/JUFE_{t}c_{i}.csv')
    process_file(odi_process,f'JUFE/JUFE.csv')
```
