```python
import os
import pandas as pd
import numpy as np
```

```python
data_all = pd.read_csv("./OIQ10K_info.csv",
                       #header = 7 ,
                       #['ref','dis','mos','dis_situation','dis_type','dis_level','source_link'],
                       encoding='utf-8',
                       sep=',')
data_distype = data_all['dis_situation']
df_cl = data_all.loc[:,['ref','dis']].replace('invalid',np.nan)
data_name = df_cl['dis'].combine_first(df_cl['ref'])
#data_name = data_all['ref'].where(data_all['ref']=='invalid',data_all['dis'])
cording = {'one distorted region':"CdistR1",
 'no perceptibly distorted region':"ConDist",
 'two distorted regions':"CdistR2",
 'global distortion':"CdistGl"}
data_reference = pd.concat([data_name,data_distype],axis=1)

data_reference = data_reference.apply(lambda x : [x['dis'].split('.')[0],x['dis_situation']],axis=1)
#data_reference['ref']=="invalid"
#print(data_reference)
data_ref = dict(data_reference.values)
print(list(data_ref.keys()))
```

```python
def find_index(idx,type_dict):
    idx_dis = dict()
    for dis in cording.values():
        idx_dis[dis] = []
    for item in idx.values:
        idx_dis[ cording[type_dict[item[0]]] ].append((item[0],item[1]))
    return idx_dis
    
for i in range(1,6):
    train = pd.read_csv(f"OIQ10K_trainc_{i}.csv",header=None,names=["ref","mos"])
    test = pd.read_csv(f"OIQ10K_testc_{i}.csv",header=None,names=["ref","mos"])
    train_dix = find_index(train,data_ref)
    test_dix = find_index(test,data_ref)
    for dis in cording.values():
        pd.DataFrame(train_dix[dis]).to_csv(f"dis_type/OIQ10K_trainc_{dis}{i}.csv",header=None,index=None)
        pd.DataFrame(test_dix[dis]).to_csv(f"dis_type/OIQ10K_testc_{dis}{i}.csv",header=None,index=None)
        
        
        
```
